````python
import numpy as np
import matplotlib as plt
import pandas as pd
#importing the datasets
df= pd.read_csv('Social_Network_Ads.csv')
````

````python
df.head()
````

````python
df.shape
````

````python
x=df.iloc[:,[2,3]]
y=df.iloc[:,4]
````

````python
x.head()
````

````python
y.head()
````

````python
#splitting data set into the training set and test set
````

````python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test =train_test_split(x,y, test_size=0.25,random_state=0)
````

````python
print("Training data:",x_train.shape)
print("Testing data:",x_test.shape)
````

````python
#feature scaling
from sklearn.preprocessing import StandardScaler
sc_x=StandardScaler()
x_train=sc_x.fit_transform(x_train)
x_test=sc_x.transform(x_test)
````

````python
from sklearn.svm import SVC
classifier=SVC(kernel='linear',random_state=0)
classifier.fit(x_train,y_train)

#predicting the test set results
y_pred =classifier.predict(x_test)
````

````python
y_pred
````

````python
from sklearn import metrics
print("Accuracy score with linear kernel:")
print(metrics.accuracy_score(y_test,y_pred))
````

````python
from sklearn.svm import SVC
classifier=SVC(kernel='rbf')
classifier.fit(x_train,y_train)
#predicting the test set results
y_pred= classifier.predict(x_test)
print("Accuracy score with rbf",metrics.accuracy_score(y_test,y_pred))
````

````python
from sklearn.svm import SVC
classifier=SVC(kernel='rbf', gamma=15,C =5,random_state=0)
classifier.fit(x_train,y_train)
#predicting the test set results
y_pred= classifier.predict(x_test)
print("Accuracy score with rbf",metrics.accuracy_score(y_test,y_pred))
````

````python
from sklearn.svm import SVC
from sklearn import metrics
svc=SVC(kernel='poly',degree=1)
svc.fit(x_train,y_train)
y_pred=svc.predict(x_test)
print("Accuracy score with poly",metrics.accuracy_score(y_test,y_pred))
````

````python
from sklearn.model_selection import GridSearchCV

# Define the parameter grid for polynomial kernel
param_grid_poly = {
    'C': [0.1, 1, 10, 100],
    'degree': [2, 3, 4],
    'gamma': ['scale', 'auto', 0.1, 0.01, 0.001],
    'coef0': [0.0, 0.1, 0.5, 1.0],
    'kernel': ['poly']
}

# Create a GridSearchCV object
grid_search_poly = GridSearchCV(estimator=SVC(random_state=0), param_grid=param_grid_poly, refit=True, verbose=2, cv=5, n_jobs=-1)

# Fit the GridSearchCV object to the training data
grid_search_poly.fit(x_train, y_train)

# Print the best parameters and best score
print("\nBest parameters found for polynomial kernel:", grid_search_poly.best_params_)
print("Best accuracy score for polynomial kernel:", grid_search_poly.best_score_)

# Make predictions with the best estimator
best_svc_poly_classifier = grid_search_poly.best_estimator_
y_pred_tuned_poly = best_svc_poly_classifier.predict(x_test)

# Evaluate the accuracy of the tuned polynomial model
print("Accuracy score with tuned polynomial kernel:", metrics.accuracy_score(y_test, y_pred_tuned_poly))
````
